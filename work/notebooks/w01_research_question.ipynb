```python
import pandas as pd
import os

# Define the repository name and the relative path to the CSV within the repository
repo_name = 'ml-internship-starter'
csv_repo_path = os.path.join('data', 'raw', 'content_refresh_anonymized.csv')

# Get the current working directory, typically '/content/' in Colab
current_colab_dir = os.getcwd()

# Construct the full path to the cloned repository root
repo_root_path = os.path.join(current_colab_dir, repo_name)

# If the repository hasn't been cloned yet, clone it
if not os.path.exists(repo_root_path):
    print(f"Cloning {repo_name} repository from GitHub...")
    # This clones the repo into the current_colab_dir (e.g., /content/)
    !git clone https://github.com/HaneefAderolu/{repo_name}.git
    print("Repository cloned.")
else:
    print(f"Repository '{repo_name}' already exists.")

# Construct the absolute path to the CSV file
full_csv_path = os.path.join(repo_root_path, csv_repo_path)

# Attempt to read the CSV file
try:
    df = pd.read_csv(full_csv_path)
    print(f"Successfully loaded data from: {full_csv_path}")
    print(f"DataFrame head:\n{df.head()}") # Adding a head print for verification
except FileNotFoundError:
    print(f"Error: The file '{full_csv_path}' was not found even after attempting to clone the repository.")
    print("Please verify the repository structure and the file's existence within it.")
    df = None # Ensure df is defined even if loading fails
except Exception as e:
    print(f"An unexpected error occurred while reading the CSV: {e}")
    df = None

print(f"Total pages: {len(df):,}")

# Signal 1: word count is the strongest correlate of impressions
corr = df[['impressions_90d','word_count','avg_position','ctr',
           'content_age_days','engagement_rate']].corr()['impressions_90d'].drop('impressions_90d')
print("\nCorrelation with impressions_90d:")
print(corr.sort_values(ascending=False).round(3))

# Signal 2: position strongly predicts CTR
pos_ctr = df.groupby(
    pd.cut(df['avg_position'], bins=[0,3,10,20,50,300],
    labels=['Top 3','4-10','11-20','21-50','50+']),
    observed=True
)['ctr'].median()
print("\nMedian CTR by position bucket:")
print(pos_ctr.round(3))

# Signal 3: pages updated 91-180 days ago get far more impressions than stale ones
fresh = df.groupby('freshness_tier', observed=True)['impressions_90d'].median()
print("\nMedian impressions by freshness tier:")
print(fresh.sort_values(ascending=False).round(0))
```
