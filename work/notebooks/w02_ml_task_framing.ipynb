## 3. Success metric

*One metric you can defend. What number means 'good'?*
Spearman rank correlation (ρ) between each signal and impressions_90d / ctr. Spearman is used instead of Pearson because the data is heavily skewed (a few pages have enormous impression counts), and Spearman is robust to outliers and non-linear relationships.

A signal is considered meaningful if |ρ| > 0.10. A signal is considered strong if |ρ| > 0.20. Anything below 0.05 is treated as noise.

The output is a ranked table of signals with their correlation values - that table is the deliverable that tells a content team where to focus.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

df = pd.read_csv('content_refresh_anonymized.csv')

# Show the unit of analysis
show_cols = ['content_id', 'impressions_90d', 'clicks_90d', 'ctr',
             'avg_position', 'word_count', 'content_age_days',
             'engagement_rate', 'trend_direction', 'trend_pct']

print("One row = one content page\n")
print(df[show_cols].head(6).to_string())

print(f"\nTotal pages: {len(df):,}")
print(f"Total signals available to analyse: {len(df.columns)} columns")

# Show the signal we care about most
numeric_signals = ['word_count','content_age_days','avg_position',
                   'engagement_rate','scroll_rate','search_volume',
                   'competition','trend_pct']

print("\nCorrelation of each signal with impressions_90d:")
corr = df[numeric_signals + ['impressions_90d']].corr()['impressions_90d'].drop('impressions_90d')
print(corr.sort_values(ascending=False).round(3).to_string())
```

## 5. Why ML beats a fixed rule here

This lane is not trying to beat a rule. It is the analysis that tells you whether rules are even possible. The finding from the data is that:

No single signal strongly predicts performance — the highest correlation in the dataset is word count at r = 0.163, which is weak
Longer pages get more impressions (3,500+ word pages have median 1,340 impressions vs 4 for pages under 1,000 words) — but this is observational
Pages ranked position 4–10 have the highest median CTR (0.16%) — better than even top-3 pages in this dataset, which is worth investigating
Content age barely correlates with performance (r = -0.001) — the common assumption that "fresher is better" is not supported here

A fixed rule like "pages over 3,500 words perform best, prioritise those" ignores all other context. This analysis shows which signals are worth including in future models (Lane 2, 4) and which are not. That is the real output: an evidence-based signal report.
